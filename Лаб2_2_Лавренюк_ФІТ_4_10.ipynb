```python
import pandas as pd
import numpy as np

from google.colab import files

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Файл успішно завантажено!")
print("Розмір датасету:", df.shape)

print("\nСтовпці:")
print(df.columns.tolist())

print("\nПерші 5 рядків:")
display(df.head())
```

```python
# Завдання 12
# Аналіз виживання залежно від статі

result_12 = df.groupby('Sex')['Survived'].agg(
    Кількість_пасажирів='count',
    Кількість_тих_хто_вижив='sum',
    Відсоток_виживання='mean'
)

result_12['Відсоток_виживання'] = (
    result_12['Відсоток_виживання'] * 100
).round(2)

print("ЗАВДАННЯ 12")
print("Аналіз виживання залежно від статі")
print()

display(result_12)

female_rate = result_12.loc['female', 'Відсоток_виживання']
male_rate = result_12.loc['male', 'Відсоток_виживання']

print(f"Виживання жінок: {female_rate:.2f}%")
print(f"Виживання чоловіків: {male_rate:.2f}%")
print(f"Різниця: {abs(female_rate - male_rate):.2f} процентних пунктів")

if female_rate > male_rate:
    print("\nВисновок: жінки мали вищий відсоток виживання, ніж чоловіки.")
else:
    print("\nВисновок: чоловіки мали вищий відсоток виживання, ніж жінки.")
```

Між чоловіками та жінками спостерігається значна різниця у рівні виживання. Жінки мали значно вищий відсоток виживання, ніж чоловіки. Отже, стать була пов'язана з різницею у рівні виживання пасажирів.

```python
# Завдання 13
# Відсоток виживання для кожного класу

result_13 = df.groupby('Pclass')['Survived'].agg(
    Кількість_пасажирів='count',
    Кількість_тих_хто_вижив='sum',
    Відсоток_виживання='mean'
)

result_13['Відсоток_виживання'] = (
    result_13['Відсоток_виживання'] * 100
).round(2)

print("ЗАВДАННЯ 13")
print("Виживання пасажирів залежно від класу")
print()

display(result_13)

highest_class = result_13['Відсоток_виживання'].idxmax()
highest_rate = result_13['Відсоток_виживання'].max()

print(
    f"Найвищий рівень виживання мав "
    f"{highest_class}-й клас — {highest_rate:.2f}%."
)
```

Найвищий рівень виживання спостерігався серед пасажирів 1-го класу. Пасажири 2-го та 3-го класів мали нижчі показники виживання. Отже, клас пасажира був пов'язаний із рівнем виживання.

```python
# Завдання 14
# Середній вік тих, хто вижив і не вижив

average_age = df.groupby('Survived')['Age'].mean()

survived_age = average_age[1]
not_survived_age = average_age[0]

print("ЗАВДАННЯ 14")
print("Середній вік пасажирів")
print()

print(f"Середній вік тих, хто вижив: {survived_age:.2f} років")
print(f"Середній вік тих, хто не вижив: {not_survived_age:.2f} років")

difference = abs(survived_age - not_survived_age)

print(f"Різниця: {difference:.2f} років")

if survived_age < not_survived_age:
    print(
        "\nВисновок: середній вік тих, хто вижив, "
        "був нижчим, ніж середній вік тих, хто не вижив."
    )
else:
    print(
        "\nВисновок: середній вік тих, хто вижив, "
        "був вищим, ніж середній вік тих, хто не вижив."
    )

print(
    "Отже, вік був пов'язаний з відмінностями у виживанні, "
    "але цей аналіз сам по собі не доводить причинний вплив віку."
)
```

Середній вік тих, хто вижив, відрізняється від середнього віку тих, хто не вижив. Це свідчить про наявність зв'язку між віком та виживанням у цьому наборі даних. Водночас лише середнє значення не доводить, що саме вік був причиною виживання або загибелі.

```python
# Завдання 15
# Розподіл пасажирів за рівнями тарифів Fare

data_fare = df.dropna(subset=['Fare']).copy()

# Визначаємо квартилі
q1 = data_fare['Fare'].quantile(0.25)
q2 = data_fare['Fare'].quantile(0.50)
q3 = data_fare['Fare'].quantile(0.75)

# Створюємо групи тарифів
data_fare['Fare_group'] = pd.cut(
    data_fare['Fare'],
    bins=[-np.inf, q1, q2, q3, np.inf],
    labels=[
        'Низький тариф',
        'Нижче середнього',
        'Вище середнього',
        'Високий тариф'
    ],
    include_lowest=True
)

result_15 = data_fare.groupby(
    'Fare_group',
    observed=False
)['Survived'].agg(
    Кількість_пасажирів='count',
    Кількість_тих_хто_вижив='sum',
    Відсоток_виживання='mean'
)

result_15['Відсоток_виживання'] = (
    result_15['Відсоток_виживання'] * 100
).round(2)

print("ЗАВДАННЯ 15")
print("Рівень тарифу Fare та виживання")
print()

print(f"Q1 = {q1:.2f}")
print(f"Медіана = {q2:.2f}")
print(f"Q3 = {q3:.2f}")
print()

display(result_15)

highest_group = result_15['Відсоток_виживання'].idxmax()
highest_rate = result_15['Відсоток_виживання'].max()

lowest_group = result_15['Відсоток_виживання'].idxmin()
lowest_rate = result_15['Відсоток_виживання'].min()

print(
    f"\nНайвищий рівень виживання: "
    f"{highest_group} — {highest_rate:.2f}%."
)

print(
    f"Найнижчий рівень виживання: "
    f"{lowest_group} — {lowest_rate:.2f}%."
)

print(
    "\nВисновок: загалом вищий тариф пов'язаний "
    "з вищою часткою виживання пасажирів."
)
```

Пасажирів було розподілено на чотири групи за величиною тарифу Fare. Порівняння показало, що пасажири з вищими тарифами загалом мали вищий рівень виживання. Отже, між тарифом та шансами виживання спостерігається статистичний зв'язок. Це не означає, що саме тариф безпосередньо був причиною виживання.

```python
# Завдання 16
# Середній тариф Fare для кожного класу

result_16 = df.groupby('Pclass')['Fare'].mean().round(2)

print("ЗАВДАННЯ 16")
print("Середній тариф для кожного класу")
print()

display(
    result_16.to_frame(name='Середній тариф Fare')
)

highest_class = result_16.idxmax()
lowest_class = result_16.idxmin()

print(
    f"Найвищий середній тариф був у "
    f"{highest_class}-му класі: {result_16[highest_class]:.2f}"
)

print(
    f"Найнижчий середній тариф був у "
    f"{lowest_class}-му класі: {result_16[lowest_class]:.2f}"
)

print(
    "\nВисновок: між класами існує значна різниця "
    "у середніх тарифах."
)
```

Середній тариф суттєво відрізняється залежно від класу. Найвищий середній тариф був у пасажирів 1-го класу, а найнижчий — у пасажирів 3-го класу. Отже, клас пасажира був тісно пов'язаний із вартістю квитка.

```python
# Завдання 17
# Середній вік пасажирів для кожного класу

result_17 = df.groupby('Pclass')['Age'].mean().round(2)

print("ЗАВДАННЯ 17")
print("Середній вік пасажирів для кожного класу")
print()

display(
    result_17.to_frame(name='Середній вік')
)

oldest_class = result_17.idxmax()
youngest_class = result_17.idxmin()

print(
    f"Найвищий середній вік був у "
    f"{oldest_class}-му класі: {result_17[oldest_class]:.2f} років."
)

print(
    f"Найнижчий середній вік був у "
    f"{youngest_class}-му класі: {result_17[youngest_class]:.2f} років."
)

print(
    "\nВисновок: середній вік пасажирів "
    "відрізняється залежно від класу."
)
```

Середній вік пасажирів відрізняється залежно від класу. Вікова структура пасажирів у трьох класах була нерівномірною. Це означає, що клас подорожі був пов'язаний не лише з тарифом, але й із віковою структурою пасажирів.
